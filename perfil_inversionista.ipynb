```python
from IPython.display import display, HTML

html_app = """
<!DOCTYPE html>
<html lang="es">
<head>
<meta charset="UTF-8">
<meta name="viewport" content="width=device-width, initial-scale=1.0">
<title>Perfil de Inversionista</title>
<style>
  * { box-sizing: border-box; margin: 0; padding: 0; }
  body {
    font-family: 'Segoe UI', system-ui, sans-serif;
    background: linear-gradient(135deg, #f0fdf8 0%, #e8f4fd 100%);
    min-height: 100vh;
    display: flex;
    align-items: flex-start;
    justify-content: center;
    padding: 2rem 1rem;
  }
  .card {
    background: #fff;
    border-radius: 20px;
    box-shadow: 0 4px 32px rgba(0,0,0,0.10);
    padding: 2rem;
    max-width: 560px;
    width: 100%;
  }
  .header { text-align: center; margin-bottom: 1.5rem; }
  .header h1 { font-size: 22px; color: #0f4c2e; font-weight: 700; margin-top: 0.5rem; }
  .header p { font-size: 14px; color: #6b7280; margin-top: 4px; }
  .progress-wrap { margin-bottom: 1.5rem; }
  .progress-label { font-size: 13px; color: #9ca3af; margin-bottom: 6px; }
  .progress-bar { height: 8px; background: #e5e7eb; border-radius: 99px; overflow: hidden; }
  .progress-fill { height: 100%; background: linear-gradient(90deg, #1D9E75, #34d399); border-radius: 99px; transition: width 0.4s ease; }
  .emoji-q { font-size: 40px; text-align: center; display: block; margin-bottom: 0.75rem; }
  .question { font-size: 18px; font-weight: 600; color: #111827; line-height: 1.5; text-align: center; margin-bottom: 1.25rem; }
  .options { display: flex; flex-direction: column; gap: 10px; }
  .opt-btn {
    display: flex; align-items: center; gap: 12px;
    padding: 14px 16px;
    border: 2px solid #e5e7eb;
    border-radius: 14px;
    background: #fff;
    cursor: pointer;
    font-size: 15px;
    color: #374151;
    text-align: left;
    transition: all 0.2s;
    width: 100%;
  }
  .opt-btn:hover { border-color: #1D9E75; background: #f0fdf8; }
  .opt-btn.selected { border-color: #1D9E75; background: #f0fdf8; color: #085041; font-weight: 600; }
  .opt-icon { font-size: 24px; min-width: 30px; text-align: center; }
  .nav { display: flex; justify-content: flex-end; margin-top: 1.5rem; }
  .btn-next {
    padding: 11px 28px;
    background: #1D9E75;
    color: #fff;
    border: none;
    border-radius: 12px;
    font-size: 15px;
    font-weight: 600;
    cursor: pointer;
    transition: opacity 0.2s, transform 0.1s;
  }
  .btn-next:disabled { opacity: 0.35; cursor: default; }
  .btn-next:not(:disabled):hover { opacity: 0.88; transform: translateY(-1px); }
  .result-emoji { font-size: 64px; text-align: center; display: block; margin-bottom: 0.75rem; }
  .result-badge {
    display: inline-block; padding: 4px 14px; border-radius: 99px;
    font-size: 12px; font-weight: 600; text-transform: uppercase;
    letter-spacing: 0.5px; margin-bottom: 0.5rem;
  }
  .result-title { font-size: 24px; font-weight: 700; color: #111827; margin-bottom: 0.75rem; text-align: center; }
  .result-desc { font-size: 15px; color: #4b5563; line-height: 1.7; text-align: center; margin-bottom: 1.25rem; }
  .score-section { margin-bottom: 1.25rem; }
  .score-labels { display: flex; justify-content: space-between; font-size: 12px; color: #9ca3af; margin-bottom: 6px; }
  .score-bar-wrap { height: 12px; background: #f3f4f6; border-radius: 99px; overflow: hidden; }
  .score-bar { height: 100%; border-radius: 99px; transition: width 0.8s ease; }
  .tip-box {
    background: #f8fafc;
    border-left: 4px solid #1D9E75;
    border-radius: 0 12px 12px 0;
    padding: 1rem 1.25rem;
    font-size: 14px;
    color: #374151;
    line-height: 1.7;
    margin-bottom: 1.5rem;
    text-align: left;
  }
  .tip-box strong { color: #085041; }
  .btn-restart {
    width: 100%;
    padding: 12px;
    background: transparent;
    color: #1D9E75;
    border: 2px solid #1D9E75;
    border-radius: 12px;
    font-size: 15px;
    font-weight: 600;
    cursor: pointer;
    transition: background 0.2s;
  }
  .btn-restart:hover { background: #f0fdf8; }
  .result-center { text-align: center; }
</style>
</head>
<body>
<div class=\"card\">
  <div class=\"header\">
    <div style=\"font-size:36px\">💰</div>
    <h1>¿Qué tipo de inversionista eres?</h1>
    <p>6 preguntas simples &bull; menos de 2 minutos</p>
  </div>

  <div id=\"quiz-screen\">
    <div class=\"progress-wrap\">
      <div class=\"progress-label\" id=\"step-label\">Pregunta 1 de 6</div>
      <div class=\"progress-bar\"><div class=\"progress-fill\" id=\"prog\" style=\"width:0%\"></div></div>
    </div>
    <span class=\"emoji-q\" id=\"q-emoji\"></span>
    <p class=\"question\" id=\"q-text\"></p>
    <div class=\"options\" id=\"options\"></div>
    <div class=\"nav\"><button class=\"btn-next\" id=\"btn-next\" disabled>Siguiente &rarr;</button></div>
  </div>

  <div id=\"result-screen\" style=\"display:none\"></div>
</div>

<script>
var questions = [
  {
    emoji: '🍬',
    text: 'Si te dan 1,000 monedas hoy, ¿qué harías con ellas?',
    opts: [
      { icon: '🏦', text: 'Las guardo en un lugar seguro y no las toco.', score: 1 },
      { icon: '🛒', text: 'Gasto un poco y ahorro el resto.', score: 2 },
      { icon: '🚀', text: 'Las invierto aunque pueda perder algunas.', score: 3 }
    ]
  },
  {
    emoji: '🎢',
    text: 'Imagina que tu dinero es una montaña rusa. ¿Cuál prefieres?',
    opts: [
      { icon: '🐌', text: 'Una muy lenta y tranquila, sin sustos.', score: 1 },
      { icon: '🎡', text: 'Con algunas subidas y bajadas, pero no tan extrema.', score: 2 },
      { icon: '😱', text: 'La más emocionante aunque dé mucho miedo.', score: 3 }
    ]
  },
  {
    emoji: '⏳',
    text: '¿Cuánto tiempo puedes esperar para ver tu dinero crecer?',
    opts: [
      { icon: '📅', text: 'Lo necesito pronto, en menos de 1 año.', score: 1 },
      { icon: '🗓️', text: 'Puedo esperar de 1 a 5 años.', score: 2 },
      { icon: '🌳', text: 'Puedo esperar más de 5 años sin problema.', score: 3 }
    ]
  },
  {
    emoji: '📉',
    text: 'Si tu dinero baja un 20% de repente, ¿qué harías?',
    opts: [
      { icon: '😰', text: 'Lo saco todo inmediatamente, no aguanto.', score: 1 },
      { icon: '🤔', text: 'Me preocupo pero espero un poco.', score: 2 },
      { icon: '😎', text: 'Aprovecho y meto más dinero.', score: 3 }
    ]
  },
  {
    emoji: '🎲',
    text: '¿Cuál juego te gusta más?',
    opts: [
      { icon: '🧩', text: 'Rompecabezas: seguro y predecible.', score: 1 },
      { icon: '♟️', text: 'Ajedrez: estrategia con algo de riesgo.', score: 2 },
      { icon: '🎲', text: 'Dados: apuesto fuerte a ganar grande.', score: 3 }
    ]
  },
  {
    emoji: '🌧️',
    text: 'Si llueve en tu día libre planeado, ¿qué haces?',
    opts: [
      { icon: '🏠', text: 'Me quedo en casa, prefiero lo seguro.', score: 1 },
      { icon: '☂️', text: 'Salgo con paraguas, me adapto.', score: 2 },
      { icon: '🌊', text: 'Salgo igual, me encanta la aventura.', score: 3 }
    ]
  }
];

var profiles = [
  {
    min: 6, max: 10,
    emoji: '🛡️',
    badge: 'Conservador',
    badgeColor: '#dcfce7',
    badgeText: '#14532d',
    barColor: '#1D9E75',
    title: 'Inversionista Conservador',
    desc: 'Prefieres la seguridad ante todo. Buscas que tu dinero esté protegido y crezcas poco a poco, sin grandes sustos. Dormirás tranquilo sabiendo que tu capital está a salvo.',
    tip: '<strong>¿Qué te conviene?</strong> Cuentas de ahorro, CDTs, bonos del gobierno y fondos de renta fija. Ganas menos pero tu dinero está muy protegido.'
  },
  {
    min: 11, max: 14,
    emoji: '⚖️',
    badge: 'Moderado',
    badgeColor: '#dbeafe',
    badgeText: '#1e3a5f',
    barColor: '#185FA5',
    title: 'Inversionista Moderado',
    desc: 'Buscas un equilibrio perfecto: algo de seguridad y algo de emoción. Aceptas subidas y bajadas pequeñas si al final ganas más que en el banco.',
    tip: '<strong>¿Qué te conviene?</strong> Fondos mixtos, acciones de empresas sólidas y finca raíz. Combinas riesgo y seguridad para crecer de forma inteligente.'
  },
  {
    min: 15, max: 18,
    emoji: '🚀',
    badge: 'Arriesgado',
    badgeColor: '#fee2e2',
    badgeText: '#7f1d1d',
    barColor: '#D85A30',
    title: 'Inversionista Arriesgado',
    desc: 'Te encanta la emoción y no le temes a las pérdidas temporales. Sabes que para ganar mucho, a veces hay que arriesgar. Tienes nervios de acero.',
    tip: '<strong>¿Qué te conviene?</strong> Acciones de crecimiento, criptomonedas, startups y mercados emergentes. Alto riesgo, pero potencial de grandes ganancias.'
  }
];

var current = 0;
var scores = [];
var selected = null;

function render() {
  var q = questions[current];
  var pct = Math.round((current / questions.length) * 100);
  document.getElementById('prog').style.width = pct + '%';
  document.getElementById('step-label').textContent = 'Pregunta ' + (current + 1) + ' de ' + questions.length;
  document.getElementById('q-emoji').textContent = q.emoji;
  document.getElementById('q-text').textContent = q.text;
  var opts = document.getElementById('options');
  opts.innerHTML = '';
  selected = null;
  document.getElementById('btn-next').disabled = true;
  q.opts.forEach(function(o) {
    var btn = document.createElement('button');
    btn.className = 'opt-btn';
    btn.innerHTML = '<span class=\"opt-icon\">' + o.icon + '</span><span>' + o.text + '</span>';
    btn.onclick = function() {
      document.querySelectorAll('.opt-btn').forEach(function(b) { b.classList.remove('selected'); });
      btn.classList.add('selected');
      selected = o.score;
      document.getElementById('btn-next').disabled = false;
    };
    opts.appendChild(btn);
  });
}

document.getElementById('btn-next').onclick = function() {
  if (selected === null) return;
  scores.push(selected);
  current++;
  if (current < questions.length) {
    render();
  } else {
    showResult();
  }
};

function showResult() {
  var total = scores.reduce(function(a, b) { return a + b; }, 0);
  var profile = profiles.find(function(p) { return total >= p.min && total <= p.max; }) || profiles[1];
  var pct = Math.round(((total - 6) / 12) * 100);
  document.getElementById('quiz-screen').style.display = 'none';
  var rs = document.getElementById('result-screen');
  rs.style.display = 'block';
  rs.innerHTML = `
    <div class=\"result-center\">
      <span class=\"result-emoji\">${profile.emoji}</span>
      <div style=\"margin-bottom:0.75rem\">
        <span class=\"result-badge\" style=\"background:${profile.badgeColor};color:${profile.badgeText}\">${profile.badge}</span>
      </div>
      <p class=\"result-title\">${profile.title}</p>
    </div>
    <div class=\"score-section\">
      <div class=\"score-labels\"><span>🛡️ Conservador</span><span>🚀 Arriesgado</span></div>
      <div class=\"score-bar-wrap\"><div class=\"score-bar\" id=\"sbar\" style=\"width:0%;background:${profile.barColor}\"></div></div>
    </div>
    <p class=\"result-desc\">${profile.desc}</p>
    <div class=\"tip-box\">${profile.tip}</div>
    <button class=\"btn-restart\" onclick=\"restart()\">🔄 Volver a empezar</button>
  `;
  setTimeout(function() {
    var bar = document.getElementById('sbar');
    if (bar) bar.style.width = pct + '%';
  }, 100);
}

function restart() {
  current = 0; scores = []; selected = null;
  document.getElementById('result-screen').style.display = 'none';
  document.getElementById('quiz-screen').style.display = 'block';
  render();
}

render();
</script>
</body>
</html>
"""

display(HTML(f'<iframe srcdoc="{html_app.replace(chr(34), "&quot;").replace(chr(39), "&#39;")}" width="100%" height="700" style="border:none; border-radius:16px;"></iframe>'))
```
